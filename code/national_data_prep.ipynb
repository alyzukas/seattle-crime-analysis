Create an `ec_cols` dataset of only the 3 economic columns we want to focus on.

I used Claude to help me with this. <br>

**Note**: the rent column has a different code in 2014.

I used Claude to help me with this. <br>
- we want to create a smaller version of each `hc_XXXX` dataset that only has the `median_rent`
- we will simutaneously add a `period` column to show the date this data came from

```python
# median gross rent: code is DP04_0132E in 2014, DP04_0134E in 2019 and 2024
hc_2014_small = hc_2014[['GEO_ID', 'NAME', 'DP04_0132E']].rename(columns={'DP04_0132E': 'median_rent'})
hc_2014_small['period'] = '2010-2014'

hc_2019_small = hc_2019[['GEO_ID', 'NAME', 'DP04_0134E']].rename(columns={'DP04_0134E': 'median_rent'})
hc_2019_small['period'] = '2015-2019'

hc_2024_small = hc_2024[['GEO_ID', 'NAME', 'DP04_0134E']].rename(columns={'DP04_0134E': 'median_rent'})
hc_2024_small['period'] = '2020-2024'

# stack them into one Housing dataset
hc_data = pd.concat([hc_2014_small, hc_2019_small, hc_2024_small], ignore_index=True)

# check it out
hc_data[hc_data['NAME'] == 'Seattle city, Washington']
```

-> Now we have 1 usable dataset for our Housing data: `hc_data`

#### Prepare the Final Census Table
Now that I have 2 datasets: `ec_data` and `hc_data`, I want to combine them into one **Census Table** <br>
- **Final table name**: `census`

##### **Merge the `ec_data` and `hc_data` datasets**

They both have the same cities and periods, so we can line them up side by side using `GEO_ID` and `period`:

```python
# merge all the columns
census = ec_data.merge(hc_data[['GEO_ID', 'period', 'median_rent']],
                       on=['GEO_ID', 'period'], how='left')

# take a look
census.head()
```

```python
# look at census data in just Seattle
census[census['NAME'] == 'Seattle city, Washington']
```

##### **Standardize Column Names**

I did not like how the `NAME` of each location was being displayed (Seattle city??) <br>
So I looked up how to change this and found that it would be too much work (I can maybe change later). <br>
So I decided to just add a `city` and `state` column at the end for now.

```python
# split "Seattle city, Washington" at the comma into place + state
census[['city', 'state']] = census['NAME'].str.rsplit(', ', n=1, expand=True)

# drop the last word ("city", "town", "CDP"...) -> "Seattle"
census['city'] = census['city'].str.rsplit(' ', n=1).str[0]

census[census['city'] == 'Seattle']
```

```python
# make all column names lowercase with underscores
census.columns = census.columns.str.lower().str.replace(' ', '_')
census.head(1)
```
